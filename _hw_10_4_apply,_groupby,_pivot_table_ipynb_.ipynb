```python
# У яких філіях частіше купують чоловіки, а в яких жінки?

hypothesis_1 = df.groupby(['Branch', 'Gender'])['Invoice ID'].count()
hypothesis_1
```

```python
hypothesis_1_perc = df.groupby('Branch')['Gender'].value_counts(normalize=True).round(4)*100
hypothesis_1_perc
```

```python
# Чи існує гендерна відмінність у виборі категорій товарів?

hypothesis_2 = df.pivot_table(index = 'Gender', columns = 'Product line', values = 'Invoice ID', aggfunc = 'count')
hypothesis_2
```

```python
hypothesis_2_perc = df.groupby('Product line')['Gender'].value_counts(normalize=True).round(4)*100
hypothesis_2_perc
```

```python
# Який спосіб оплати є найпопулярнішим серед клієнтів, і чи відрізняється цей вибір у звичайних покупців та членів клубу?

hypothesis_3 = df.pivot_table(index = 'Customer type', columns = 'Payment', values = 'Invoice ID', aggfunc = 'count')
hypothesis_3
```

```python
hypothesis_3_perc = df.groupby('Customer type')['Payment'].value_counts(normalize=True).round(4)*100
hypothesis_3_perc
```
