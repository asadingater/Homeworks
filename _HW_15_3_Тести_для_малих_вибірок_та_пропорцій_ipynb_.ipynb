# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


```python
import numpy as np
from scipy import stats

data = np.array([128, 135, 121, 142, 126, 151, 123])
mu_0 = 123.7

print("1. Гіпотези та тип тесту")
print("H0: mu = 123.7 (середній час не змінився)")
print("Ha: mu > 123.7 (середній час збільшився)")
print("Тип тесту: Правосторонній t-тест для малої вибірки\n")

print("2. Вибіркові статистики")
x_bar = np.mean(data)
s = np.std(data, ddof=1)
n = len(data)
df = n - 1
print(f"Вибіркове середнє (x_bar): {x_bar:.3f}")
print(f"Стандартне відхилення (s): {s:.3f}")
print(f"Розмір вибірки (n): {n}")
print(f"Ступені свободи (df): {df}\n")

print("3. Виконання тесту (alpha = 0.10)")
SE = s / np.sqrt(n)
t_stat = (x_bar - mu_0) / SE

t_crit_10 = stats.t.ppf(1 - 0.10, df=df)

p_value = 1 - stats.t.cdf(t_stat, df=df)

print(f"Статистика тесту (t): {t_stat:.3f}")
print(f"Критичне значення t (alpha=0.10): {t_crit_10:.3f}")
print(f"p-значення (p-value): {p_value:.4f}\n")

print("4. Прийняття рішення")
if p_value < 0.10:
    print(f"Оскільки p-value ({p_value:.4f}) < alpha (0.10), ми ВІДХИЛЯЄМО нульову гіпотезу H0.")
else:
    print(f"Оскільки p-value ({p_value:.4f}) >= alpha (0.10), ми НЕ ВІДХИЛЯЄМО нульову гіпотезу H0.")
print()

print("5. Висновок")
print("Редизайн сайту дійсно статистично подовжив середній час до покупки на рівні значущості 10%.\n")

print("6. Зміна рівня значущості на alpha = 0.05")
t_crit_05 = stats.t.ppf(1 - 0.05, df=df)
print(f"Критичне значення t (alpha=0.05): {t_crit_05:.3f}")
if p_value < 0.05:
    print(f"Рішення НЕ зміниться. Оскільки p-value ({p_value:.4f}) < 0.05, ми все одно ВІДХИЛЯЄМО H0.")
else:
    print(f"Рішення ЗМІНИТЬСЯ. Оскільки p-value ({p_value:.4f}) >= 0.05, ми НЕ зможемо відхилити H0.")
```

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

count = 311
nobs = 1200
p_0 = 0.23
alpha = 0.03

z_stat, p_value = proportions_ztest(count, nobs, value=p_0, alternative='larger')

ci_low, ci_high = proportion_confint(count, nobs, alpha=alpha, method='normal')

print(f"\n1. СТАДІЯ ФОРМУЛЮВАННЯ ГІПОТЕЗ:")
print(f" - H0: p = 0.23 (впізнаваність логотипу не змінилася)")
print(f" - Ha: p > 0.23 (впізнаваність логотипу зросла — правосторонній тест)")

print(f"\n2. ОБЧИСЛЕНІ РОЗРАХУНКОВІ ПОКАЗНИКИ:")
print(f" - Вибіркова пропорція (p_hat): {count/nobs:.4f} ({count/nobs*100:.2f}%)")
print(f" - Розрахована Z-статистика:   {z_stat:.4f}")
print(f" - Отримане p-value:           {p_value:.4f}")
print(f" - Довірчий інтервал ({int((1-alpha)*100)}%):    ({ci_low:.4f}, {ci_high:.4f})")

print(f"\n3. ПРИЙНЯТТЯ РІШЕННЯ ЩОДО H0:")
if p_value < alpha:
    print(f" - Оскільки p-value ({p_value:.4f}) < alpha ({alpha}), ми відхиляємо нульову гіпотезу H0.")
else:
    print(f" - Оскільки p-value ({p_value:.4f}) >= alpha ({alpha}), ми не відхиляємо нульову гіпотезу H0.")

print(f"\n4. ФІНАЛЬНИЙ ВИСНОВОК:")
print(f" - Маркетингові дані дають достатні підстави стверджувати, що після проведення спеціальної рекламної кампанії впізнаваність логотипу компанії статистично значущо зросла.")

print(f"\n5. ІНТЕРПРЕТАЦІЯ ДОВІРЧОГО ІНТЕРВАЛУ:")
print(f" - Зі встановленою надійністю {int((1-alpha)*100)}% справжня нова частка людей, які впізнають логотип, лежить у межах від {ci_low*100:.2f}% до {ci_high*100:.2f}%.")
print(f" - Як це пояснює рішення: Нижній поріг довірчого інтервалу ({ci_low*100:.2f}%) є повністю більшим за початкові 23.00%. Це означає, що навіть за найменш оптимістичного сценарію нова впізнаваність перевищує старий рівень, що додатково підтверджує відхилення H0. ")
```
